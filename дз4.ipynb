```python
import gradio as gr
from datetime import datetime

def application(receiver, full_name, position, from_where, to_where, sign_name):
    current_date = datetime.now().strftime("%d.%m.%Y")

    text = f"""
                                          {receiver}

                          ЗАЯВА
           ПРО ПЕРЕВЕДЕННЯ ДО ІНШОГО ПІДРОЗДІЛУ

Прошу перевести мене, {full_name},
на посаду '{position}'
з '{from_where}' до '{to_where}'.

Дата: {current_date}.                   Підпис: ___________
                                          {sign_name}
    """
    return text

demo = gr.Interface(
    fn=application,
    inputs=[
        gr.Textbox(label="Кому (ПІБ керівника)", placeholder="Директору мережі лікарень (їхні ПІБ(у давальному відмінку))"),
        gr.Textbox(label="Ваше ПІБ (у родовому відмінку)"),
        gr.Textbox(label="Нова посада (у родовому відмінку)"),
        gr.Textbox(label="З якого підрозділу: (у родовому відмінку)"),
        gr.Textbox(label="В який підрозділ: (у родовому відмінку)"),
        gr.Textbox(label="Ваше ПІБ (у називному відмінку)"),
    ],
    outputs=gr.Textbox(label="Текст заяви", placeholder="Зовсім скоро тут з'явиться ваша готова заява!"),
    title="Генератор заяв про переведення",
    description="Заповніть поля нижче, щоб сформувати заяву."
)

if __name__ == "__main__":
    demo.launch()
```
