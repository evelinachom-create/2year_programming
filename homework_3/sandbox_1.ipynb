```python
# 3. Гра з монеткою
# Монета має 2 грані - орел та решка. Ми підкидаємо монету вгору і отримуємо рандомний результат.
#Виведіть результат підкидання та порахуйте, скільки разів випаде орел та решка, якщо ми підкинемо монету 100 разів? 1000?

import random

def flip_coin(times):
    sides = (["Орел", "Решка"])

    results = [random.choice(sides) for _ in range(times)]

    heads = results.count("Орел")
    tails = results.count("Решка")

    print(f"Підкинули монету {times} разів.")
    print(f"Орел випав {heads} разів.")
    print(f"Решка випала {tails} разів.")

flip_coin(100)
print("-" * 35)
flip_coin(1000)


```

```python
# 4*. Створіть свій рандомайзер-гру (камінь-ножиці-папір, лотерея, кидок кубиків, будь-яка інша).

import random

def start_game():
    secret_number = random.randint(1, 30)
    attempts = 5
    
    print("Гра: Вгадай число")
    print("*" * 35)
    print("Було загадане число від 1 до 30.")
    print(f"У вас є {attempts} спроб, щоб вгадати його.\n")
    
    for attempt in range(1, attempts + 1):
        guess = int(input(f"Спроба {attempt}/{attempts}. Введіть ваше число: "))
        
        if guess == secret_number:
            print(f"Ура! Ви вгадали! Правильне число: {secret_number}")
            break

        if attempt < attempts:
            if guess < secret_number:
                print("Спробуйте ввести більше число.")
            else:
                print("Спробуйте ввести менше число.")

        print("-" * 35)
        
    else:
        print(f"\n Спроби вичерпано! Ви програли. Загадане число: {secret_number}")

start_game()
```
