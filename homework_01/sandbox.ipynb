```python
# 1) Створіть власний виняток
# Створити клас:
# class EmptyTextError(Exception):
# pass
# Написати функцію:
# def analyze_text(text):
# ...
# Функція повинна:
# перевірити, чи є text рядком;
# перевірити, чи не є текст порожнім;
# якщо текст порожній — створити EmptyTextError;
# якщо текст правильний — повернути кількість символів.

class EmptyTextError(Exception):
    pass

def analyze_text(text):
    if not isinstance(text, str):
        raise TypeError("Введене має бути лише рядком.")
    if text == "":
        raise EmptyTextError("Текст не має бути порожнім.")
    
    return len(text)

try:
    print(analyze_text("Код на Python для перевірки тексту"))
    print(analyze_text("Hello, world!"))
    print(analyze_text(""))

except (EmptyTextError, TypeError) as e:
    print(f"Помилка: {e}")
```

```python
# 2) Користувач вводить регулярний вираз:
# Введіть регулярний вираз:
# Програма повинна: отримати регулярний вираз; скомпілювати його за допомогою re.compile(); перехопити re.error; повідомити користувача про некоректний вираз.
# Наприклад, користувач вводить: [A-Z
# Програма повинна повідомити:  Помилка: регулярний вираз має некоректний синтаксис.
# Якщо регулярний вираз правильний, попросити користувача ввести текст і виконати пошук збігів.

import re

def compile_re_search_match():
    pattern_str = input("Введіть регулярний вираз: ")

    try:
        pattern = re.compile(pattern_str)
    except re.error:
        print("Помилка: регулярний вираз має некоректний синтаксис.")
        return

    text_input = input("Введіть текст для пошуку збігів: ")
    matches = pattern.findall(text_input)

    if matches:
        print(f"Збіги знайдено: {matches}")
    else:
        print("Жодного збігу не знайдено.")

compile_re_search_match()
```

```python
# 3) Написати програму, яка читає файл: text55.txt і виводить його вміст.

# Передбачити обробку:
# FileNotFoundError;
# PermissionError;
# UnicodeDecodeError.

# Додати підрахунок: кількості символів; кількості слів; кількості рядків.

def analyze_textfile(filename = "text55.txt"):
    try:
        with open(filename, "r", encoding="utf-8") as file:
            content = file.read()

    except FileNotFoundError:
        print("Помилка: вказаний файл не знайдено.")

    except PermissionError:
        print("Помилка: немає прав доступу до файлу.")

    except UnicodeDecodeError:
        print("Помилка: не вдалося декодувати вміст файлу.")

    print(content)

    print(f"Кількість символів: {len(content)}")
    print(f"Кількість слів: {len(content.split())}")
    print(f"Кількість рядків: {len(content.splitlines())}")


analyze_textfile()
```
